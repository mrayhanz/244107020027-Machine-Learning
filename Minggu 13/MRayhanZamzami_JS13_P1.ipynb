# PRAKTIKUM 1 - Backpropagation Manual

```python
import numpy as np

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

np.random.seed(42)
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

for epoch in range(10001):
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    error = y - a2

    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 2000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.5f}")

print("Prediksi:")
print(a2)
```

```python
# Tugas 1: hidden_size = 3
hidden_size_3 = 3

np.random.seed(42)
W1_3 = np.random.randn(input_size, hidden_size_3)
b1_3 = np.zeros((1, hidden_size_3))
W2_3 = np.random.randn(hidden_size_3, output_size)
b2_3 = np.zeros((1, output_size))

for epoch in range(10001):
    z1 = np.dot(X, W1_3) + b1_3
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2_3) + b2_3
    a2_3 = sigmoid(z2)

    error = y - a2_3

    d_a2 = error * sigmoid_derivative(a2_3)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_3.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_3 += lr * d_W1
    b1_3 += lr * d_b1
    W2_3 += lr * d_W2
    b2_3 += lr * d_b2

    if epoch % 2000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss (hidden 3): {loss:.5f}")

print("Prediksi (hidden 3):")
print(a2_3)
```

```python
# Tugas 1: Aktivasi ReLU pada hidden layer
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

np.random.seed(42)
W1_r = np.random.randn(input_size, 4) * np.sqrt(2.0 / input_size)
b1_r = np.zeros((1, 4))
W2_r = np.random.randn(4, output_size) * np.sqrt(2.0 / 4)
b2_r = np.zeros((1, output_size))

lr_relu = 0.05
for epoch in range(10001):
    z1 = np.dot(X, W1_r) + b1_r
    a1 = relu(z1)
    z2 = np.dot(a1, W2_r) + b2_r
    a2_r = sigmoid(z2)

    error = y - a2_r

    d_a2 = error * sigmoid_derivative(a2_r)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_r.T) * relu_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_r += lr_relu * d_W1
    b1_r += lr_relu * d_b1
    W2_r += lr_relu * d_W2
    b2_r += lr_relu * d_b2

    if epoch % 2000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss (ReLU): {loss:.5f}")

print("Prediksi (ReLU):")
print(a2_r)
```

# TUGAS PRAKTIKUM - Klasifikasi Angka MNIST Menggunakan ANN

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
import time

(X_train, y_train), (X_test, y_test) = mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0

y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10, batch_size=64, verbose=1)

loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Akurasi Baseline pada data uji: {acc:.4f}")
```

```python
# Eksperimen 1: 256 dan 128 Neuron
m1 = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(256, activation='relu'),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])
m1.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
t0 = time.time()
m1.fit(X_train, y_train, epochs=5, batch_size=64, verbose=0)
t1 = time.time() - t0
_, acc1 = m1.evaluate(X_test, y_test, verbose=0)
print(f"Neuron (256, 128) - Waktu: {t1:.2f}s, Akurasi: {acc1:.4f}")
```

```python
# Eksperimen 2: Tambah 1 Hidden Layer (3 Hidden Layers)
m2 = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(256, activation='relu'),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
m2.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
t0 = time.time()
m2.fit(X_train, y_train, epochs=5, batch_size=64, verbose=0)
t2 = time.time() - t0
_, acc2 = m2.evaluate(X_test, y_test, verbose=0)
print(f"3 Hidden Layers (256, 128, 64) - Waktu: {t2:.2f}s, Akurasi: {acc2:.4f}")
```

```python
# Eksperimen 3: Sigmoid vs ReLU
m_sig = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='sigmoid'),
    Dense(64, activation='sigmoid'),
    Dense(10, activation='softmax')
])
m_sig.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
t0 = time.time()
m_sig.fit(X_train, y_train, epochs=5, batch_size=64, verbose=0)
t_sig = time.time() - t0
_, acc_sig = m_sig.evaluate(X_test, y_test, verbose=0)
print(f"Aktivasi Sigmoid - Waktu: {t_sig:.2f}s, Akurasi: {acc_sig:.4f}")
print(f"Aktivasi ReLU    - Akurasi Baseline: {acc:.4f}")
```

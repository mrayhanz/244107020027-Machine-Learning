# PRAKTIKUM 1 - Klasifikasi Cat vs Dog Menggunakan CNN

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

train_dir = 'dataset/training_set' if os.path.exists('dataset/training_set') else '../dataset/training_set'
test_dir = 'dataset/test_set' if os.path.exists('dataset/test_set') else '../dataset/test_set'

train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)
training_set = train_datagen.flow_from_directory(
    train_dir,
    target_size=(64, 64),
    batch_size=32,
    class_mode='binary'
)

test_datagen = ImageDataGenerator(rescale=1./255)
test_set = test_datagen.flow_from_directory(
    test_dir,
    target_size=(64, 64),
    batch_size=32,
    class_mode='binary'
)
```

```python
cnn = tf.keras.models.Sequential()
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=[64, 64, 3]))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
cnn.add(tf.keras.layers.Flatten())
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))

cnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
cnn.summary()
```

```python
history = cnn.fit(x=training_set, validation_data=test_set, epochs=10)
```

```python
import numpy as np
from tensorflow.keras.preprocessing import image

single_pred_path = 'dataset/single_prediction/cat_or_dog_1.jpg' if os.path.exists('dataset/single_prediction/cat_or_dog_1.jpg') else '../dataset/single_prediction/cat_or_dog_1.jpg'
test_image = image.load_img(single_pred_path, target_size=(64, 64))
test_image = image.img_to_array(test_image)
test_image = np.expand_dims(test_image, axis=0) / 255.0

result = cnn.predict(test_image)
print("Class Indices:", training_set.class_indices)
prediction = 'dog' if result[0][0] > 0.5 else 'cat'
print("Hasil Prediksi Single Image:", prediction, f"(Probability: {result[0][0]:.4f})")
```

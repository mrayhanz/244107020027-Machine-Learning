```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns

dpath = 'dataset/pbl/diabetes_binary_5050split_health_indicators_BRFSS2015.csv'
if not os.path.exists(dpath):
    dpath = '../dataset/pbl/diabetes_binary_5050split_health_indicators_BRFSS2015.csv'

df = pd.read_csv(dpath)

from sklearn.decomposition import PCA
from sklearn.cluster import HDBSCAN
from sklearn.preprocessing import StandardScaler

X = df.drop('Diabetes_binary', axis=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

clusterer = HDBSCAN(min_cluster_size=50, min_samples=10)
labels = clusterer.fit_predict(X_pca)

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', s=5)
plt.title('HDBSCAN Clustering (PCA Projection)')
plt.show()
```

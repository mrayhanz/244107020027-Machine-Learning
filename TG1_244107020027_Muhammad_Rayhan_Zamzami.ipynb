Fungsi Tiap Library yang Digunakan
1. mne: Library yang digunakan untuk mengolah dan menganalisis data sinyal otak, khususnya data dari EEG (Electroencephalography) dan MEG (Magnetoencephalography).
2. pyPREP: Library yang berfungsi membantu proses preprocessing pada data EEG dengan menerapkan metode PREP (Preprocessing Pipeline) untuk mempersiapkan data sebelum masuk ke tahap analisis lebih lanjut.
3. scipy: Library Python yang menyediakan berbagai fungsi untuk komputasi ilmiah dan pengolahan data. Beberapa kegunaannya meliputi pemrosesan sinyal, analisis statistik, aljabar linear, optimasi, interpolasi, serta transformasi Fourier.
4. wandb: Library yang digunakan untuk memantau dan mengelola eksperimen machine learning maupun deep learning. Library ini dapat mencatat metrik selama proses training, membuat visualisasi hasil, menyimpan konfigurasi hyperparameter, serta membantu membandingkan beberapa eksperimen atau model.
5. pyECG: Library yang digunakan untuk mengakses dan mengelola data ECG (Electrocardiogram), yaitu data berupa rekaman aktivitas listrik yang dihasilkan oleh jantung.
